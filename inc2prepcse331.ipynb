```python
peng = pd.read_csv('penguins.csv')

# using assign for silly shit

peng['bill_ratio'] = peng['bill_length_mm'] / peng['bill_depth_mm']
peng = peng.assign(bill_ratio1 = peng['bill_length_mm'] / peng['bill_depth_mm'])
print(peng[['bill_ratio', 'bill_ratio1']][:5])


# Using np.select

conditions = [
    peng['body_mass_g'].isna(),
    peng['body_mass_g'] < 3500,
    (peng['body_mass_g'] >= 3500) & (peng['body_mass_g'] < 4500),
    peng['body_mass_g'] >= 4500
]

values = ['missing', 'low', 'medium', 'high']

peng['mass_category'] = np.select(conditions, values, default = 'unknown')
print(peng['mass_category'][:5])


peng = peng.assign(
    body_mass_g = lambda df: df.body_mass_g.fillna(df.body_mass_g.mean())
)



peng['mass_category_2'] = np.select(conditions, values, default = 'unknown')
print(peng[['mass_category', 'mass_category_2']][:5])
```

```python
new_data = peng.groupby('species', as_index = False).agg(
    mean_bill_length = ('bill_length_mm', 'mean'),
    median_bill_depth = ('bill_depth_mm', 'median'),
    n = ('species', 'count')
).sort_values(by = 'n', ascending = False)
print(new_data)
```

```python
summary = [
    peng
    .dropna(subset = ['bill_length_mm', 'body_mass_g', 'bill_depth_mm'])
    .assign(bill_ratio = lambda df: df['bill_length_mm'] / df['bill_depth_mm'])
    .groupby('species', as_index = False)
    .agg(mean_bill_length = ('bill_length_mm', 'mean'), median_bill_depth = ('bill_depth_mm', 'median'), count = ('species', 'count'))
    .sort_values(by = 'count', ascending=False)
]   
print(summary)
```

```python

df = pd.DataFrame({
    "id": [1, 2, 3],
    "math": [80, 85, 90],
    "english": [70, 78, 88]
})



useless_long = pd.melt(df)
print(useless_long)

healthy_long = df.melt(
    id_vars = 'id',
    value_vars= ['math', 'english'],
    var_name = 'subject',
    value_name = 'score'
)
print("=" * 50)
print(healthy_long)
```

```python
wider = healthy_long.pivot(
    index = 'id',
    columns = 'subject',
    values = 'score'
).reset_index(inplace = False)
print(wider)
```

```python
students = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "name": ["A", "B", "C", "D"]
})
scores = pd.DataFrame({
    "id": [2, 3, 3, 5],
    "score": [80, 90, 85, 70]
})



inner_join = pd.merge(left = students, right = scores, how =  'inner', on  = 'id')
print(inner_join)


full_join = pd.merge(left = students, right = scores, how = 'outer', on  = 'id')
print(full_join)

left_join = students.merge(scores, how = 'left', on = 'id')
print(left_join)

right_join = students.merge(scores, how = 'right', left_on = 'id', right_on = 'id', suffixes = ('_sc', '_st'))
print(right_join)
```
